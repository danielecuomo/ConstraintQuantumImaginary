```python
import numpy as np
from scipy.optimize import minimize

Hm = H.to_matrix()

constraints = [
    SparsePauliOp(["XII"]).to_matrix(),
    SparsePauliOp(["IXI"]).to_matrix(),
    SparsePauliOp(["IIX"]).to_matrix(),
]

targets = np.array([
    target,
    target,
    target
])

dim = Hm.shape[0]


def normalize(x):

    n = np.linalg.norm(x)

    if n < 1e-14:
        return None

    return x / n


def energy(x):

    psi = normalize(x)

    return np.vdot(
        psi,
        Hm @ psi
    ).real


#
# equality constraints
#

cons = []

#
# norm = 1
#

cons.append({
    "type": "eq",
    "fun": lambda x:
        np.linalg.norm(x)**2 - 1.0
})

#
# <C_k> = target_k
#

for Ck, tk in zip(
    constraints,
    targets
):

    cons.append({
        "type": "eq",
        "fun": lambda x, Ck=Ck, tk=tk:
            np.vdot(
                normalize(x),
                Ck @ normalize(x)
            ).real - tk
    })


best_E = np.inf
best_psi = None

rng = np.random.default_rng(1234)

for trial in range(50):

    x0 = rng.normal(
        size=dim
    )

    x0 /= np.linalg.norm(x0)

    result = minimize(
        energy,
        x0,
        method="SLSQP",
        constraints=cons,
        options={
            "maxiter": 10000,
            "ftol": 1e-12
        }
    )

    if result.success:

        E = energy(result.x)

        if E < best_E:

            best_E = E

            best_psi = normalize(
                result.x
            )

print()
print("best energy =", best_E)

print()
print("constraints")

for i, Ck in enumerate(
    constraints
):

    value = np.vdot(
        best_psi,
        Ck @ best_psi
    ).real

    print(
        i,
        value,
        "target=",
        targets[i]
    )
```

```python
import numpy as np


def pauli_to_symplectic(label):

    n = len(label)

    x = np.zeros(
        n,
        dtype=np.uint8
    )

    z = np.zeros(
        n,
        dtype=np.uint8
    )

    for i, p in enumerate(label):

        if p == "X":
            x[i] = 1

        elif p == "Z":
            z[i] = 1

        elif p == "Y":
            x[i] = 1
            z[i] = 1

    return x, z


def op_to_symplectic(op):

    #
    # attenzione:
    # qui assumiamo che ogni operatore
    # della base sia una singola stringa Pauli
    #
    label = op.paulis[0].to_label()

    x, z = pauli_to_symplectic(
        label
    )

    return np.concatenate(
        [x, z]
    )


def gf2_rank(M):

    M = (
        M.copy()
        .astype(np.uint8)
    )

    nrows, ncols = M.shape

    rank = 0
    col = 0

    while (
        rank < nrows
        and col < ncols
    ):

        pivot = None

        for r in range(
            rank,
            nrows
        ):

            if M[r, col]:
                pivot = r
                break

        if pivot is None:
            col += 1
            continue

        M[[rank, pivot]] = (
            M[[pivot, rank]]
        )

        for r in range(nrows):

            if (
                r != rank
                and M[r, col]
            ):
                M[r] ^= M[rank]

        rank += 1
        col += 1

    return rank


#
# build symplectic matrix
#

Ssym = np.vstack([
    op_to_symplectic(
        basis.operator(i)
    )
    for i in range(
        basis.size
    )
])

print(
    "number of operators =",
    basis.size
)

print(
    "symplectic rank =",
    gf2_rank(Ssym)
)

print(
    "max possible =",
    2 * len(
        basis.operator(0)
        .paulis[0]
        .to_label()
    )
)
```

```python
diffs = []

for vin, vout in unique_pairs:

    vin = np.array(vin, dtype=np.uint8)
    vout = np.array(vout, dtype=np.uint8)

    diffs.append(
        np.bitwise_xor(vin, vout)
    )

#
# rimuovi duplicati
#
diffs = np.unique(
    np.array(diffs),
    axis=0
)

print("generators:")
print(diffs)
```

```python
def gf2_rank(M):

    M = M.copy().astype(np.uint8)

    nrows, ncols = M.shape

    rank = 0
    col = 0

    while rank < nrows and col < ncols:

        pivot = None

        for r in range(rank, nrows):

            if M[r, col]:
                pivot = r
                break

        if pivot is None:
            col += 1
            continue

        M[[rank, pivot]] = M[[pivot, rank]]

        for r in range(nrows):

            if r != rank and M[r, col]:
                M[r] ^= M[rank]

        rank += 1
        col += 1

    return rank


r = gf2_rank(diffs)

print("rank =", r)
print("reachable states =", 2**r)
```

```python
import numpy as np
from itertools import product


#
# generatori simplessici di H
#

generators = []

for p in H_eff.paulis:

    x, z = pauli_to_symplectic(
        p.to_label()
    )

    generators.append(
        np.concatenate([x, z]) % 2
    )

generators = np.array(
    generators,
    dtype=np.uint8
)

r = len(generators)

#
# tutti i vettori simplessici
# presenti nella closure
#

closure = set()

for i in range(basis.size):

    op = basis.operator(i)

    for p in op.paulis:

        x, z = pauli_to_symplectic(
            p.to_label()
        )

        closure.add(
            tuple(
                np.concatenate([x, z]) % 2
            )
        )

print(
    "closure size =",
    len(closure)
)

#
# scegliamo un seed
# (primo Pauli della closure)
#

v0 = np.array(
    next(iter(closure)),
    dtype=np.uint8
)

#
# costruzione affine
# v0 + span(H)
#

affine = set()

for coeffs in product(
    [0, 1],
    repeat=r
):

    v = v0.copy()

    for c, g in zip(
        coeffs,
        generators
    ):

        if c:
            v ^= g

    affine.add(
        tuple(v)
    )

print(
    "affine size =",
    len(affine)
)

#
# confronto
#

missing = affine - closure
extra   = closure - affine

print(
    "missing from closure =",
    len(missing)
)

print(
    "extra in closure =",
    len(extra)
)
```
