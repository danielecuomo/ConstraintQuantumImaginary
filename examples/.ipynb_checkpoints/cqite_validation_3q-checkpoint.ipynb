```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import expm
from qiskit.quantum_info import Statevector, SparsePauliOp
from qitp.engine.unitary_approximation import UnitaryApproximation
from qitp.engine.propagation import QITEPropagation
from qitp.engine.cqite_run import run_qite
from qitp.engine.prepare_state import constraint_rotation, prepare_state
from qitp.simulation.backend import SimulatorBackend
```

```python
from qitp.ansatz.pauli_basis import (
    PauliBasis
)
from qitp.ansatz.commutator_basis import (
    CommutatorBasis
)
```

```python
from qiskit.quantum_info import SparsePauliOp

h = 1.0
J = 1.0

H = SparsePauliOp(
    [
        "XII",
        "IXI",
        "IIX",
        "ZZI",
        "ZIZ",
        "IZZ",
    ],
    coeffs=[h, h, h, J, J, J]
)
```

```python
Mz = constraints = [
    SparsePauliOp(["ZII"], coeffs=[1.0]),
    SparsePauliOp(["IZI"], coeffs=[1.0]),
    SparsePauliOp(["IIZ"], coeffs=[1.0]),
]

Identity = SparsePauliOp(["III"],coeffs=[1.0])
```

```python
seeds = [
    SparsePauliOp("ZII"),
    SparsePauliOp("IZI"),
    SparsePauliOp("IIZ"),
    SparsePauliOp("XII"),
    SparsePauliOp("IXI"),
    SparsePauliOp("IIX")
]
```

```python
basis = CommutatorBasis.from_seed(hamiltonian=H, seeds=seeds, depth=1)
```

```python
basis.size
```

```python
approx = UnitaryApproximation(basis)

backend = SimulatorBackend()

prop = QITEPropagation(
    H,
    approx,
    backend
)

prop.constraints = [
    Identity]
prop.constraints.extend(Mz)
```

```python
prop.constraints
```

```python
dt=1e-1
nsteps=30
target=0.5

energies=[]
fidelities=[]
errors=[]
magnetizations=[]
```

```python
rotations = []

for constraint in Mz:

    rotations.append(
        constraint_rotation(
            constraint,
            target
        )
    )
```

```python
circuit = prepare_state(
    3,
    rotations
)
```

```python
circuit = circuit.decompose()
```

```python
circuit.draw("mpl")
```

```python
results = run_qite(
    prop,
    circuit,
    H,
    Mz,
    dt,
    nsteps
)

energies = results["energies"]
magnetizations = results["magnetizations"]
```

```python
print(np.round(energies[-1],3))
```

```python
plt.plot(energies); plt.title('Energy'); plt.grid(); plt.show()
```

```python
plt.plot(magnetizations); plt.title("Magnetization"); plt.grid(); plt.show()
```

```python
evals = np.linalg.eigvalsh(
    H.to_matrix()
)

print("Exact spectrum")

print(np.round(evals,2))
```

```python
import numpy as np
from scipy.optimize import root_scalar

Hm = H.to_matrix()
Mm = Mz.to_matrix()

def mz_expectation(lam):

    Hc = Hm + lam * Mm

    E, V = np.linalg.eigh(Hc)

    psi = V[:, 0]

    return np.real(
        np.vdot(
            psi,
            Mm @ psi
        )
    )


sol = root_scalar(
    lambda lam:
        mz_expectation(lam) - target,
    bracket=[-100, 100]
)

lam = sol.root

Hc = Hm + lam * Mm

E, V = np.linalg.eigh(Hc)

psi_exact = V[:, 0]

E_exact = np.real(
    np.vdot(
        psi_exact,
        Hm @ psi_exact
    )
)

print("E_exact =", np.round(E_exact,2))
```
