```python
from qiskit import QuantumCircuit
from qitp.hamiltonians.pauli_hamiltonian import PauliHamiltonian, PauliString
from qitp.engine.propagation import QITEPropagation
from qitp.engine.unitary_approximation import UnitaryApproximation
from qitp.simulation.backend import SimulatorBackend
from qiskit.quantum_info import SparsePauliOp
```

```python
H = PauliHamiltonian([
    PauliString("ZZ", pi/4),
    PauliString("IX", pi/4),
    PauliString("XI", pi/4)
])

# Define a Pauli basis for the approximation
pauli_basis = [
    SparsePauliOp("XX"),
    SparsePauliOp("ZZ"),
]

approx = UnitaryApproximation(pauli_basis)

backend = SimulatorBackend()

Mz = PauliHamiltonian([
    PauliString("ZI", 1.0),
    PauliString("IZ", 1.0)
])

engine = QITEPropagation(H, approx, backend)

engine.constraints = [Mz]
engine.targets = [1.0]   # target magnetization

initial_circuit = QuantumCircuit(2)

#initial_circuit.h(0)
#initial_circuit.h(1)
theta = 2*np.arccos(np.sqrt(3/4))
initial_circuit.ry(theta,0)
initial_circuit.ry(theta,1)


state = backend.run(initial_circuit)

dt = 0.1
e0 = inf
steps = 25

import math
best = math.inf
best_state = None

for _ in range(steps):

    state = engine.step(state, dt)
    

    energy = H.expectation(state).real

    if energy < best:
        best = energy
        best_state = state

    #print(energy)

    #for i, Bi in enumerate(engine.constraints):
    #    val = state.expectation_value(Bi.operator).real
    #    print(f"Constraint {i}: {val:.2e} (target {engine.targets[i]})")
```
